Dans le cas présent je vais repartir les modalités en chiffres selon les hypoithèses suivantes : 
- 1 pourrait représenter un paiement unique (équivalent à "Yearly").
- 4 pourrait représenter une fréquence trimestrielle (équivalent à "Quarterly").
- 6 pourrait représenter une fréquence semestrielle (équivalent à "Biannual").
- 12 pourrait représenter une fréquence mensuelle (équivalent à "Monthly").


La méthode va consister à replace chaque chiffre par son équivalent

Nous allons filter la base pour voir la cohérence des varibles drv_drv2, drv_sex1, drv_sex2, drv_age1, drv_age2, drv_age_lic1 et drv_age_lic2

```python
# Filtrer les lignes où drv_drv2 est 'Yes'
filtered_df = base_ssDoub[base_ssDoub['drv_drv2'] == 'Yes']

# Filtrer les lignes où drv_sex1 est NaN, drv_sex2 est 'M' ou 'F', et drv_age1, drv_age2, drv_age_lic1 et drv_age_lic2 ont des valeurs
result_df = filtered_df[
    (filtered_df['drv_sex1'].isna()) &
    (filtered_df['drv_sex2'].isin(['M', 'F'])) &
    (filtered_df['drv_age1'].notnull()) &
    (filtered_df['drv_age2'].notnull()) &
    (filtered_df['drv_age_lic1'].notnull()) &
    (filtered_df['drv_age_lic2'].notnull())
]

# Afficher le résultat
result_df
```

On constate qu'il y a environ 123 valeurs manquantes qui respectent les conditions du filtre ce qui nous rassure que nos valeurs manquantes de la variable drv_sex1 doivent être trouver  

Nous allons utilisé la méthode de proportions pour repartir les 123 valeurs manquantes de la variable drv_sex1


```python
# Calculer les proportions
proportions = base_ssDoub['drv_sex1'].value_counts(normalize=True)
proportions = proportions.dropna()  # Exclure les NaN
```

```python
# Remplir les NaN aléatoirement en fonction des proportions
base_ssDoub['drv_sex1'] = base_ssDoub['drv_sex1'].apply(
    lambda x: np.random.choice(proportions.index, p=proportions.values) if pd.isna(x) else x
)
```

On constate que l'identifant claimb change pour certains clients ayant le même identifiant et le même identifaint de vehicule, on ne peut donc pas les supprimer. 

```python
# Supprimer le préfixe "amount= " de la variable claim_amount
pg['claim_amount'] = pg['claim_amount'].str.replace('amount= ', '')
# Convertir la colonne claim_amount en numérique
pg['claim_amount'] = pd.to_numeric(pg['claim_amount'], errors='coerce')
# Afficher les premières lignes pour vérifier le résultat
pg.head()
```

```python
pg[pg['id_client'].duplicated()].sort_values(by='id_client').head(50)
```

```python
other_columns = [col for col in pg.columns if col not in ['id_client', 'id_vehicle','claim_amount', 'claim_nb']] 
df_sin = pg[['id_client', 'id_vehicle','claim_amount', 'claim_nb']]
aggsin = df_sin.groupby(['id_client','id_vehicle']).agg(
    nb_sin=pd.NamedAgg(column='claim_nb', aggfunc='sum'),  # Compter le nombre de sinistre par client
    remb_total=pd.NamedAgg(column='claim_amount', aggfunc='sum'),  # Somme des montants des réclamations par client
).reset_index()

aggsin.head(10)
```
